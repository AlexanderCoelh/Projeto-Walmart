```python
#colocando nome em cada categoria de acordo com a empresa
map_dept = {
    38: 'Mercearia Seca',
    72: 'Higiene Pessoal',
    40: 'Eletronicos',
    2: 'Produtos de Limpeza',
    95: 'Padaria e Confeitaria',
    5: 'Carnes e Congelados',
    13: 'Laticinios',
    23: 'Hortifruti',
    10: 'Pet Shop',
    7: 'Utilidades Domésticas'
}

df_loja_3['Category'] = df_loja_3['Dept'].map(map_dept)
```

```python
df_loja_3_dept = df_loja_3[df_loja_3['Dept'].isin([38, 72, 40, 2, 95, 5, 13, 23, 10, 7])] #selecionando apenas as categorias que são as mais vendidas
```

```python
df_loja_3_dept.groupby('IsHoliday')['Weekly_Sales'].mean()

#Ao analisar as vendas em semanas com feriados, é notavel que semanas que contém feriados tem resultados superiores as que não tem
```

```python
df_loja_3_dept.groupby('IsHoliday')['Weekly_Sales'].median() #verificando possiveis outliers, percebe-se que não influenciaram na Análise
```

```python
df_loja_3_dept['Date'] = pd.to_datetime(df_loja_3_dept['Date']) #convertendo para datatime
df_loja_3_dept.info()
```

```python
df_loja_3_dept['Year'] = df_loja_3_dept['Date'].dt.year #criando a coluna Year, para analisar por ano
df_loja_3_dept['Month'] = df_loja_3_dept['Date'].dt.month_name()
df_loja_3_dept
```

Ao analisar o ano de 2012, observa-se que esse período apresentou menor
variabilidade das vendas semanais quando comparado a 2010 e 2011, indicando
um comportamento mais estável ao longo do ano. Além disso, 2012 registrou
a maior média e a maior mediana de vendas semanais para a loja 3, o que
sugere um desempenho médio superior, com menor influência de eventos
externos extremos.

Entretanto, apesar desse desempenho médio mais elevado, o volume total de
vendas em 2012 foi inferior aos anos anteriores, o que se explica pelo
menor número de semanas contabilizadas no dataset. Dessa forma, a análise
reforça a importância de considerar simultaneamente métricas de tendência
central, dispersão e volume ao interpretar os resultados.

```python
df_loja_3_dept.groupby('Year')['Weekly_Sales'].sum().sort_values(ascending=False)
```

```python
df_loja_3_dept['Year'].value_counts()
```

```python
df_loja_3_dept.groupby('Year')['Weekly_Sales'].mean().sort_values(ascending= False)
```

```python
df_loja_3_dept.groupby('Year')['Weekly_Sales'].median().sort_values(ascending= False)
```

```python
df_loja_3_dept.groupby('Year')['Weekly_Sales'].std()
```

Apesar de julho ser um dos meses com maior número de registros no sistema,
seu desempenho é inferior aos demais, pois as vendas concentram-se em
categorias de baixo valor agregado, como mercearia seca e produtos de consumo
recorrente.

Em contrapartida, dezembro apresenta alto desempenho, impulsionado por
categorias de maior valor agregado, comportamento fortemente associado a
datas comemorativas e compras sazonais.

Observa-se também que novembro apresenta elevado desvio padrão, indicando
alta variabilidade nas vendas semanais. Esse comportamento pode ser explicado
pela Black Friday, que ocorre na última sexta-feira do mês, gerando picos de
vendas pontuais.

Esses resultados sugerem que o comportamento do consumidor é fortemente
influenciado por preço e eventos promocionais. Estratégias como a criação de
kits, ações promocionais e o estímulo ao cross-sell podem contribuir para o
aumento do ticket médio e melhorar o desempenho dos meses mais fracos.

```python
#Contabilizando a quantidade de registros de cada mês

df_loja_3_dept['Month'].value_counts()
```

```python
# Calcula métricas mensais para comparar volume e desempenho médio

df_loja_3_dept.groupby('Month')['Weekly_Sales'].sum().sort_values(ascending=False)
```

```python
#Calculando a mediana de cada mês

df_loja_3_dept.groupby('Month')['Weekly_Sales'].median().sort_values(ascending= False)
```

```python
#Observando o desvio padrão referente aos meses registrados

df_loja_3_dept.groupby('Month')['Weekly_Sales'].std().sort_values(ascending= False)
```
